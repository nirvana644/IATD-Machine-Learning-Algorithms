# 1. NumPy Arrays

NumPy arrays come in two forms : vectors and matrices. Vectors are strictly 1-dimensional (1D) arrays and matrices are 2D (but you should note a matrix can still have only one row or one column).

#### Why use Numpy array? Why not just a list?

*Please revise your python basics before you start understanding these libraries and their concepts.

There are lot's of reasons to use a Numpy array instead of a "standard" python list object. 
Our main reasons are:
* Memory Efficiency of Numpy Array when compared to python lists
* Easily expnadable to n-dimensions
* Speed of calculations of numpy array* 
* Numpy is an integral library for all the data science and machine learning projects

## Task 1.6 Random 
With Numpy you can create random number arrays:

### rand
Creates an array of the given shape and populates it with random samples from a uniform distribution over ``[0, 1)``. [[reference](https://docs.scipy.org/doc/numpy-1.15.0/reference/generated/numpy.random.rand.html)]

```python
np.random.rand(3)
```

```python
np.random.rand(4,4)
```

### randn

Returns a sample (or samples) from the "standard normal" distribution [σ = 1]. Unlike **rand** which is uniform, values closer to zero are more likely to appear. [[reference](https://docs.scipy.org/doc/numpy-1.15.0/reference/generated/numpy.random.randn.html)]

```python
np.random.randn(3)
```

```python
np.random.randn(4,4)
```

### randint
Returns random integers from `low` (inclusive) to `high` (exclusive).  [[reference](https://docs.scipy.org/doc/numpy-1.15.0/reference/generated/numpy.random.randint.html)]

```python
np.random.randint(1,10)
```

```python
np.random.randint(1,10,5)
```

### seed
Can be used to set the random state, so that the same "random" results can be reproduced. [[reference](https://docs.scipy.org/doc/numpy-1.15.0/reference/generated/numpy.random.seed.html)]

```python
np.random.seed(15)
np.random.rand(3)
```
