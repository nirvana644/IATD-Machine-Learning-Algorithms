## Task 1.2 Built-in Methods to create arrays

There are lots of built-in ways to generate arrays.

### arange

Return evenly spaced values within a given interval. [[reference — deprecated](https://docs.scipy.org/doc/numpy-1.15.0/reference/generated/numpy.arange.html)]. Check the range carefully for inclusion and exclusion.
[updated reference](https://numpy.org/doc/stable/reference/generated/numpy.arange.html)

`arange` can be called with a varying number of positional arguments:

- `arange(stop)`: Values are generated within the half-open interval `[0, stop)` (in other words, the interval including start but excluding stop).
- `arange(start, stop)`: Values are generated within the half-open interval `[start, stop)`.
- `arange(start, stop, step)`: Values are generated within the half-open interval `[start, stop)`, with spacing between values given by step.


```python
np.arange(10,20)
```

```python
np.arange(10,20,2)
```

### Task 1.4 linspace 
Return evenly spaced numbers over a specified interval. [[reference](https://www.numpy.org/devdocs/reference/generated/numpy.linspace.html)]

`numpy.linspace(start, stop, num=50, endpoint=True, retstep=False, dtype=None, axis=0, *, device=None)`

Parameters:
- start: array_like

  The starting value of the sequence.
- stop: array_like

  The end value of the sequence, unless endpoint is set to False. In that case, the sequence consists of all but the last of num + 1 evenly spaced samples, so that stop is excluded. Note that the step size changes when endpoint is False.
- num: int, optional

  Number of samples to generate. Default is 50. Must be non-negative.

```python
np.linspace(0,15,4)
```

## Task 1.6 Random 
With Numpy you can create random number arrays:

### rand
Creates an array of the given shape and populates it with random samples from a uniform distribution over ``[0, 1)``. [[reference — deprecated](https://docs.scipy.org/doc/numpy-1.15.0/reference/generated/numpy.random.rand.html)]

[updated reference](https://numpy.org/doc/stable/reference/random/generated/numpy.random.rand.html)

`random.rand(d0, d1, ..., dn)`

Creates an array of the given shape and populates it with random samples from a uniform distribution over ``[0, 1)``.

```python
np.random.rand(3)
```

```python
values = np.random.rand(4, 4)

print(f"Type of values: {type(values)}")
print(values)
print('Average:', values.mean())
```

Example with more dimensions:

```python
np.random.rand(4,4,4)
```

### randn

Returns a sample (or samples) from the "standard normal" distribution [σ = 1]. Unlike **rand** which is uniform, values closer to zero are more likely to appear. [[reference — deprecated](https://docs.scipy.org/doc/numpy-1.15.0/reference/generated/numpy.random.randn.html)]

[updated reference](https://numpy.org/doc/stable/reference/random/generated/numpy.random.randn.html)

```python
np.random.randn(3)
```

```python
values = np.random.randn(4,4)

#print(f"Type of values: {type(values)}")
print(f"Type of values: {type(values).__module__}.{type(values).__name__}")
print(values)
print(f"Shape of values: {values.shape}")
print(f"Size of values: {values.size}") # the ndarray.size attribute returns the total number of elements in an array
print(f"Average of values: {values.mean()}")
```

Comparing `rand` and `randn`:

```python
print('rand average:', np.random.rand(1000, 1000).mean())
print('randn average:', np.random.randn(1000, 1000).mean())
```

### randint
Returns random integers from `low` (inclusive) to `high` (exclusive).  [[reference — deprecated](https://docs.scipy.org/doc/numpy-1.15.0/reference/generated/numpy.random.randint.html)]

[updated reference](https://numpy.org/doc/stable/reference/random/generated/numpy.random.randint.html)

`random.randint(low, high=None, size=None, dtype=int)`
Return random integers from low (inclusive) to high (exclusive).

Return random integers from the “discrete uniform” distribution of the specified dtype in the “half-open” interval [low, high). If high is None (the default), then results are from [0, low).

```python
np.random.randint(1,10)
```

```python
np.random.randint(1,10,5)
```

### seed
Can be used to set the random state, so that the same "random" results can be reproduced. [[reference — deprecated](https://docs.scipy.org/doc/numpy-1.15.0/reference/generated/numpy.random.seed.html)]

[updated reference](https://numpy.org/doc/stable/reference/random/generated/numpy.random.seed.html)
This is a convenience, **legacy function** that exists to support older code that uses the singleton RandomState. Best practice is to use a dedicated `Generator` instance rather than the random variate generation methods exposed directly in the random module.

```python
np.random.seed(15)
np.random.rand(3)
```

It can be observed that the same random numbers are generated each time the code is run. This is useful for reproducibility of results, especially in scientific experiments or simulations where you want to ensure that the same random numbers are used across different runs of the code.

## Reshape
In arr array we have 20 values. In this example, we will reshape 20 values in 4 rows and 5 columms.
Returns an array containing the same data with a new shape. [[reference — deprecated](https://docs.scipy.org/doc/numpy-1.15.0/reference/generated/numpy.reshape.html)]

[updated reference](https://numpy.org/doc/stable/reference/generated/numpy.reshape.html)

`numpy.reshape(a, /, shape, order='C', *, copy=None)`

### Parameters

- **`a`** — array-like

  Input array to reshape.

- **`shape`** — integer or tuple of integers

  Specifies the output dimensions, which must accommodate all input elements.
  An integer produces a one-dimensional array. One dimension may be `-1`,
  allowing NumPy to calculate it.

- **`order`** — `'C'`, `'F'`, or `'A'`, optional

  Sets the indexing order used to read and place elements:

  - `'C'`: last axis varies fastest; the default.
  - `'F'`: first axis varies fastest.
  - `'A'`: follows Fortran order for Fortran-contiguous input;
    otherwise, follows C order.

  This controls indexing, not the output's memory layout.

- **`copy`** — boolean or `None`, optional

  - `True`: always copies the data.
  - `None`: copies only when necessary; the default.
  - `False`: raises `ValueError` if reshaping requires copying.

### Returns

An `ndarray` with the requested shape: a view when possible, otherwise a copy.
Its memory layout is not guaranteed.

```python
arr.reshape(4,5)
```
