# The Sigmoid Function Clearly Explained

The sigmoid function is defined as

$$ σ(x) = \frac{1}{1 + e^{-x}}$$

It is denoted by the Greek letter *sigma*.

Reference: [`numpy.exp`](https://numpy.org/doc/stable/reference/generated/numpy.exp.html)

Calculate the exponential of all elements in the input array.

```python
import matplotlib.pyplot as plt
import numpy as np


# Define the sigmoid function
def sigmoid(x):
  return 1 / (1 + np.exp(-x))


# Generate x values from -10 to 10
x = np.linspace(-10, 10, 100)
y = sigmoid(x)

# Plot the S-shaped curve
plt.figure(figsize=(8, 4))
plt.plot(x, y, label=r"$\sigma(x) = \frac{1}{1 + e^{-x}}$", color="blue")
plt.axhline(0, color="gray", linestyle="--", linewidth=0.5)
plt.axhline(1, color="gray", linestyle="--", linewidth=0.5)
plt.axvline(0, color="red", linestyle=":", linewidth=1)
plt.title("Sigmoid Activation Function")
plt.xlabel("Input ($x$)")
plt.ylabel(r"Output ($\sigma(x)$)")
plt.grid(True)
plt.legend()
plt.hlines(y=0.5, xmin=-1, xmax=1, color="green", linewidth=2)
plt.text(1.2, 0.5, "0.5", color="green", va="center")
plt.hlines(y=1, xmin=-1, xmax=1, color="green", linewidth=2)
plt.text(1.2, 1, "1", color="green", va="center")
plt.show()
```

Exponential functions are functions of the form $f(x) = a^x$, where $a$ is a positive constant. The sigmoid function is a special case of an exponential function.

Let us plot:

$$ e^{-x} = \frac{1}{e^{x}} $$

```python
x = np.linspace(-2, 2, 100)
y_exp = np.exp(x)
y_exp_negative = np.exp(-x)

# Plot the exponential functions
plt.figure(figsize=(8, 4))
plt.plot(x, y_exp, label=r"$e^x$", color="blue")
plt.plot(x, y_exp_negative, label=r"$e^{-x}$", color="orange")
plt.axhline(0, color="gray", linestyle="--", linewidth=0.5)
plt.axhline(1, color="gray", linestyle="--", linewidth=0.5)
plt.axvline(0, color="red", linestyle=":", linewidth=1)
plt.title("Exponential Functions")
plt.xlabel("Input ($x$)")
plt.ylabel("Output")
plt.grid(True)
plt.legend()
plt.show()
```

It can be seen how it grows towards infinity, and converges to 1 on x=0:

$$
\begin{aligned}
\lim_{x \to -\infty} e^{-x} &= +\infty \\
\lim_{x \to 0} e^{-x} &= 1 \\
\lim_{x \to \infty} e^{-x} &= 0
\end{aligned}
$$

## Limits of the Sigmoid

$$
\begin{aligned}
\lim_{x \to -\infty} σ(x) = \lim_{x \to -\infty} \frac{1}{1 + e^{-x}} = \frac{1}{1 +\infty} = 0 \\
\lim_{x \to 0} σ(x) = \lim_{x \to 0} \frac{1}{1 + e^{-x}} = \frac{1}{1 + 1} = 0.5 \\
\lim_{x \to \infty} σ(x) = \lim_{x \to \infty} \frac{1}{1 + e^{-x}} = \frac{1}{1 +0} = 1 \\
\end{aligned}
$$

The sigmoid function has a **squashing effect**, mapping any input value to a value between 0 and 1. This makes it useful for binary classification problems, where the output can be interpreted as the probability of the input belonging to a certain class.

What this means is:

$$ -\infty \leq x \leq \infty \Rightarrow 0 < σ(x) < 1$$

For any value that x takes, the value of the sigmoid function is always between zero and one.

$$ x \in \mathbb{R} \Rightarrow 0 < \sigma(x) < 1 $$
