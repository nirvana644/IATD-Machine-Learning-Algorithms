The sigmoid function is defined as

$$ σ(x) = \frac{1}{1 + e^{-x}}$$

It is denoted by the Greek letter *sigma*.

Reference: [`numpy.exp`](https://numpy.org/doc/stable/reference/generated/numpy.exp.html)

Calculate the exponential of all elements in the input array.

```python
import matplotlib.pyplot as plt
import numpy as np


# Define the sigmoid function
def sigmoid(x):
  return 1 / (1 + np.exp(-x))


# Generate x values from -10 to 10
x = np.linspace(-10, 10, 100)
y = sigmoid(x)

# Plot the S-shaped curve
plt.figure(figsize=(8, 4))
plt.plot(x, y, label=r"$\sigma(x) = \frac{1}{1 + e^{-x}}$", color="blue")
plt.axhline(0, color="gray", linestyle="--", linewidth=0.5)
plt.axhline(1, color="gray", linestyle="--", linewidth=0.5)
plt.axvline(0, color="red", linestyle=":", linewidth=1)
plt.title("Sigmoid Activation Function")
plt.xlabel("Input ($x$)")
plt.ylabel(r"Output ($\sigma(x)$)")
plt.grid(True)
plt.legend()
plt.hlines(y=0.5, xmin=-1, xmax=1, color="green", linewidth=2)
plt.text(1.2, 0.5, "0.5", color="green", va="center")
plt.hlines(y=1, xmin=-1, xmax=1, color="green", linewidth=2)
plt.text(1.2, 1, "1", color="green", va="center")
plt.show()
```

Exponential functions are functions of the form $f(x) = a^x$, where $a$ is a positive constant. The sigmoid function is a special case of an exponential function.

Let us plot:

$$ e^{-x} = \frac{1}{e^{x}} $$

```python
x = np.linspace(-2, 2, 100)
y_exp = np.exp(x)
y_exp_negative = np.exp(-x)

# Plot the exponential functions
plt.figure(figsize=(8, 4))
plt.plot(x, y_exp, label=r"$e^x$", color="blue")
plt.plot(x, y_exp_negative, label=r"$e^{-x}$", color="orange")
plt.axhline(0, color="gray", linestyle="--", linewidth=0.5)
plt.axhline(1, color="gray", linestyle="--", linewidth=0.5)
plt.axvline(0, color="red", linestyle=":", linewidth=1)
plt.title("Exponential Functions")
plt.xlabel("Input ($x$)")
plt.ylabel("Output")
plt.grid(True)
plt.legend()
plt.show()
```

It can be seen how it grows towards infinity, and converges to 1 on x=0:

$$
\begin{aligned}
\lim_{x \to -\infty} e^{-x} &= +\infty \\
\lim_{x \to 0} e^{-x} &= 1 \\
\lim_{x \to \infty} e^{-x} &= 0
\end{aligned}
$$

## Limits of the Sigmoid

$$
\begin{aligned}
\lim_{x \to -\infty} σ(x) = \lim_{x \to -\infty} \frac{1}{1 + e^{-x}} = \frac{1}{1 +\infty} = 0 \\
\lim_{x \to 0} σ(x) = \lim_{x \to 0} \frac{1}{1 + e^{-x}} = \frac{1}{1 + 1} = 0.5 \\
\lim_{x \to \infty} σ(x) = \lim_{x \to \infty} \frac{1}{1 + e^{-x}} = \frac{1}{1 +0} = 1 \\
\end{aligned}
$$

The sigmoid function has a **squashing effect**, mapping any input value to a value between 0 and 1. This makes it useful for binary classification problems, where the output can be interpreted as the probability of the input belonging to a certain class.

What this means is:

$$ -\infty \leq x \leq \infty \Rightarrow 0 < σ(x) < 1$$

For any value that x takes, the value of the sigmoid function is always between zero and one.

$$ x \in \mathbb{R} \Rightarrow 0 < \sigma(x) < 1 $$

**Why this illustration uses two sets of coefficients:** In the example created with the assistant, the coefficients were chosen separately to give a clear visual comparison resembling the reference figure:

- Linear model: $b_0 = -0.1$, $b_1 = 0.12$.
- Logistic model: $b_0 = -5$, $b_1 = 1$.

These illustrative values were not fitted to data. They make both curves pass through $(5, 0.5)$, while showing a gradual blue line and a steeper red S-shaped curve. Although the labels use the same generic symbols $b_0$ and $b_1$, their values differ between the two models in this plot. Therefore, the red curve is not the sigmoid of the displayed blue line.

To show the direct transformation instead, use one shared set of coefficients: calculate the linear score $z = b_0 + b_1x$, then calculate $p = \sigma(z) = 1 / (1 + e^{-z})$. The example below headed "Using the same coefficients" demonstrates this relationship.

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 10, 500)

# Linear model
b0_linear = -0.1
b1_linear = 0.12
y = b0_linear + b1_linear * x

# Logistic model
b0_logistic = -5
b1_logistic = 1
p = 1 / (1 + np.exp(-(b0_logistic + b1_logistic * x)))

fig, ax = plt.subplots(figsize=(10, 5))

ax.plot(x, y, color="royalblue", linewidth=3,
        label=r"Linear: $y = b_0 + b_1x$")
ax.plot(x, p, color="crimson", linewidth=3,
        label=r"Logistic: $p = \frac{1}{1 + e^{-(b_0 + b_1x)}}$")

# Reference lines and axes
ax.axhline(1, color="lightgray", linewidth=2)
ax.spines["bottom"].set_position(("data", 0))
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)

for side in ["left", "bottom"]:
    ax.spines[side].set_color("midnightblue")
    ax.spines[side].set_linewidth(2)

ax.set_xlim(0, 10)
ax.set_ylim(-0.2, 1.2)
ax.set_yticks([0, 0.5, 1])
ax.set_xlabel("$x$", loc="right", fontsize=14)
ax.set_ylabel("$y$, $p$", fontsize=14)
ax.set_title("Linear and Logistic Models")
ax.legend(loc="upper left")

plt.tight_layout()
plt.show()
```

Using the same coefficients:

```python
b0 = -5
b1 = 1

linear_score = b0 + b1 * x
p = 1 / (1 + np.exp(-linear_score))

plt.plot(x, linear_score, color="blue", label="Linear score")
plt.plot(x, p, color="red", label="Logistic probability")
```
