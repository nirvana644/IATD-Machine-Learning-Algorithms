# About GridSearchCV

Dipanjan stated that the number of options tried is the factorial of the sum of the options. But that is not true.

[Grid Search](https://scikit-learn.org/stable/modules/grid_search.html)

*specifies that two grids should be explored: one with a linear kernel and C values in [1, 10, 100, 1000], and the second one with an RBF kernel, and the **cross-product** of C values ranging in [1, 10, 100, 1000] and gamma values in [0.001, 0.0001].*

## Using Numpy

```python
from itertools import product

param_grid = {
    "max_depth": [3, 5, 10],
    "min_samples_leaf": [2, 5, 10],
    "criterion": ["gini", "entropy"]
}

combinations = list(product(*param_grid.values()))

print(len(combinations))
```

## Using Scikit-Learn

```python
from sklearn.model_selection import ParameterGrid
import pandas as pd

combinations = list(ParameterGrid(param_grid))

display(pd.DataFrame(combinations))
```

## Comparing Numpy and Scikit-Learn with the factorial of the sum

```python
from math import factorial, prod

options = [3, 3, 2]

cartesian_product = prod(options)
factorial_of_sum = factorial(sum(options))

print(f"Cartesian product: {cartesian_product:,}")
print(f"Factorial of the sum: {factorial_of_sum:,}")
print(f"Ratio: {factorial_of_sum / cartesian_product:,.0f}")
```
