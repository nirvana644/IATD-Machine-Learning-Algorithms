# DataFrames

Throughout the course, most of our data exploration will be done with DataFrames. DataFrames are an extremely powerful tool and a natural extension of the Pandas Series. By definition all a DataFrame is:

**A Pandas DataFrame consists of multiple Pandas Series that share index values.**

```python
import numpy as np
import pandas as pd
np.random.seed(101)
data = np.random.randint(0,50,(5,4))
```

```python
data
```

```python
rows = ['SYD', 'MEL', 'CAN', 'BRI', 'PER']
```

```python
columns = ['JAN', 'FEB', 'MAR', 'APR']
```

```python
df = pd.DataFrame(data=data)
df
```

```python
df = pd.DataFrame(data=data,index=rows)
df
```

```python
df = pd.DataFrame(data=data,columns=columns)
df
```

```python
df = pd.DataFrame(data=data,index=rows,columns=columns)
df
```

<h4> Creating a new column for the Dataframe </h4>

```python
df['goal_difference'] = df['goals_scored'] - df['goals_conceded']
```

```python
df.head()
```

```python
df['total_goals'] = df['goals_scored'] + df['goals_conceded']
```

```python
df.head()
```

<h4> Manipulating columns using Numpy </h4>

```python
# Because pandas is based on numpy, we get awesome capabilities with numpy's universal functions!
df['attendance_thousands'] = np.round(df['attendance'] / 1000, 2)
```

<h4>Removing columns from the data frame </h4>

```python
df = df.drop("goal_difference",axis=1)
```
