## Worked example: `loc` vs `iloc`

Both select rows and columns using square brackets: `example.loc[rows, columns]` or `example.iloc[rows, columns]`. Omitting the column selector selects all columns.

| Feature | `loc` | `iloc` |
|---|---|---|
| Selects using | Index and column **labels** | Integer **positions**, starting at 0 |
| Single value | `example.loc['SYD', 'FEB']` | `example.iloc[0, 1]` |
| Slice endpoint | Ending label is **included** | Ending position is **excluded** |
| Meaning of an integer | A label, if the index contains integers | A position |

The small, fictional dataset below uses the city and month labels from earlier in this notebook. It is self-contained and uses `indexing_example` so the existing `df` is preserved.

```python
import pandas as pd
from IPython.display import display

indexing_example = pd.DataFrame(
    {'JAN': [10, 20, 30, 40],
     'FEB': [11, 21, 31, 41],
     'MAR': [12, 22, 32, 42]},
    index=['SYD', 'MEL', 'CAN', 'BRI']
)
display(indexing_example)
```

### 1. Select a row or a single value

`SYD` is the label of the first row, at position `0`. `FEB` is the label of the second column, at position `1`. Each pair below selects the same data.

For this unique index, `loc['SYD']` and `iloc[0]` return a **Series**. Passing a list instead — `loc[['SYD']]` and `iloc[[0]]` — keeps the result as a **one-row DataFrame**, which displays as a table in PyCharm. The extra brackets do not select extra data; they preserve the two-dimensional structure. To view the Series as plain text, use `print(indexing_example.loc['SYD'])`.

Selecting one row and one column returns a scalar value.

```python
print('Row selected by label (one-row DataFrame):')
display(indexing_example.loc[['SYD']])
print('Same row selected by position (one-row DataFrame):')
display(indexing_example.iloc[[0]])

print('SYD / FEB using loc:', indexing_example.loc['SYD', 'FEB'])
print('SYD / FEB using iloc:', indexing_example.iloc[0, 1])  # Both return 11
```

### 2. Select a range of rows and columns

With `loc`, `'SYD':'CAN'` includes `SYD`, `MEL`, and `CAN`; `'JAN':'FEB'` includes both columns. With `iloc`, `0:3` selects row positions 0, 1, and 2, and `0:2` selects column positions 0 and 1. The two results below are identical.

Label slices follow the existing index order; they do not sort city names alphabetically.

```python
print('loc: includes the ending labels CAN and FEB')
display(indexing_example.loc['SYD':'CAN', 'JAN':'FEB'])

print('iloc: excludes ending positions 3 and 2')
display(indexing_example.iloc[0:3, 0:2])
```

### 3. Integer labels are not necessarily positions

Here we give a copy integer row labels `[10, 20, 0, 30]`. The row **labelled `0`** is now the third row, while the row at **position `0`** is still the first row. Consequently, `loc[0]` and `iloc[0]` return different rows. Below we use `loc[[0]]` and `iloc[[0]]` to keep those results as one-row DataFrames for table display.

This distinction matters after filtering or reordering a DataFrame: labels can remain unchanged while positions change.

```python
integer_labels_example = indexing_example.copy()
integer_labels_example.index = [10, 20, 0, 30]
display(integer_labels_example)

print('loc[[0]]: row with label 0 (the third row, JAN = 30)')
display(integer_labels_example.loc[[0]])

print('iloc[[0]]: row at position 0 (the first row, JAN = 10)')
display(integer_labels_example.iloc[[0]])
```

**Remember:** use `loc` when you know the row/column names, and `iloc` when you know their positions. `loc` includes the last label in a slice; `iloc` stops before the last position.
