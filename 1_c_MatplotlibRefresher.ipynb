First we plot our target curve prior to adding the projections. The curve is defined parametrically by the following equations:
\begin{align}
x &= r \sin(\theta) \\
y &= r \cos(\theta) \\
z &= z
\end{align}

````python
ax = plt.figure().add_subplot(projection='3d')

# Prepare arrays x, y, z
theta = np.linspace(-4 * np.pi, 4 * np.pi, 100)
z = np.linspace(-2, 2, 100)
r = z**2 + 1
x = r * np.sin(theta)
y = r * np.cos(theta)

ax.plot(x, y, z, label='parametric curve')
ax.legend()

plt.show()
````

The code below plots the parametric 2D projections of the curve onto the XY and YZ planes. The 2D plots are created using subplots, which allows for multiple plots to be displayed in a single figure. The aspect ratio of each subplot is set to 'equal' to ensure that the scales of the axes are the same, and a grid is added for better visualization.

````python
fig, axes = plt.subplots(1, 2, figsize=(10, 4), layout='constrained')

axes[0].plot(x, y, color='blue')
axes[0].set(xlabel='X', ylabel='Y', title='XY projection')

axes[1].plot(y, z, color='orange')
axes[1].set(xlabel='Y', ylabel='Z', title='YZ projection')

for ax in axes:
    ax.set_aspect('equal', adjustable='box')
    ax.grid(alpha=0.3)

plt.show()
````

Finally we plot the parametric curve in 3D and overlay the 2D projections onto the main plot. The 3D plot is created using the `projection='3d'` argument in `add_axes`, and the 2D plots are added as smaller axes on top of the main plot. The `zorder` parameter is used to control the layering of the plots, ensuring that the 2D projections are displayed above the 3D plot.

````python
fig = plt.figure(figsize=(9, 7))

# Prepare arrays x, y, z
theta = np.linspace(-4 * np.pi, 4 * np.pi, 100)
z = np.linspace(-2, 2, 100)
r = z**2 + 1
x = r * np.sin(theta)
y = r * np.cos(theta)

# Main 3D plot
axes1 = fig.add_axes([0.05, 0.05, 0.85, 0.85], projection='3d')
axes1.plot(x, y, z) # The label is redundant: label='Parametric curve'
axes1.set(
    xlabel='X', ylabel='Y', zlabel='Z',
    #title='Parametric curve'
)
axes1.set_title(
    'Parametric curve',
    color='crimson',
    fontweight='bold',
    fontsize=14,
)
#axes1.legend(loc='lower left')
axes1.xaxis.label.set_color('orangered')
axes1.yaxis.label.set_color('orangered')
axes1.zaxis.label.set_color('orangered')

# Smaller 2D plot overlaid on the main plot
axes2 = fig.add_axes([0.65, 0.65, 0.28, 0.28], zorder=10)
axes2.set_facecolor('white')
axes2.plot(x, y, color='blue')
axes2.set(xlabel='X', ylabel='Y') #, title='XY projection')
axes2.set_title(
    'XY projection',
    color='blue',
    fontweight='normal',
    fontsize=10,
)
axes2.set_aspect('equal', adjustable='box')
axes2.grid(alpha=0.3)
axes2.tick_params(axis='both', labelcolor='blue')
axes2.xaxis.label.set_color('blue')
axes2.yaxis.label.set_color('blue')


# Smaller 2D plot overlaid on the main plot
axes3 = fig.add_axes([0.0, 0.15, 0.28, 0.28], zorder=10)
axes3.set_facecolor('white')
axes3.plot(y, z, color='orange')
axes3.set(xlabel='Y', ylabel='Z') #, title='YZ projection')
axes3.set_title(
    'YZ projection',
    color='orange',
    fontweight='normal',
    fontsize=10
)
axes3.set_aspect('equal', adjustable='box')
axes3.grid(alpha=0.3)
axes3.tick_params(axis='both', labelcolor='orange')
axes3.xaxis.label.set_color('orange')
axes3.yaxis.label.set_color('orange')

# XY projection border
for spine in axes2.spines.values():
    spine.set_edgecolor('blue')
    spine.set_linewidth(2)

# YZ projection border
for spine in axes3.spines.values():
    spine.set_edgecolor('orange')
    spine.set_linewidth(2)

axes2.set_facecolor('aliceblue')
axes3.set_facecolor('cornsilk')

plt.show()

fig.savefig('./images/week_1/figure_parametric.png',bbox_inches='tight')
````

Using the example I did before with the parametric curve, lets see how we can use subplots to create a similar figure with the main 3D plot and the two 2D projections.

| Arguments | Meaning |
|---|---|
| `1, 3, 1` | 1 row, 3 columns, left subplot |
| `1, 3, 2` | 1 row, 3 columns, middle subplot |
| `1, 3, 3` | 1 row, 3 columns, right subplot |

Subplot positions start at **1**, while your Python list indices (`axes[0]`, etc.) start at **0**.

````python
fig = plt.figure(figsize=(15, 5), layout='constrained')
# add_subplot(nrows, ncols, index): rows, columns, and subplot position.
# Positions start at 1 and run left to right, then top to bottom.
# Here the layout is one row with three columns: [1] [2] [3].
axes = [
    fig.add_subplot(1, 3, 1, projection='3d'),  # Position 1 (left): 3D curve
    fig.add_subplot(1, 3, 2),  # Position 2 (middle): XY projection
    fig.add_subplot(1, 3, 3),  # Position 3 (right): YZ projection
]

# Prepare arrays x, y, z
theta = np.linspace(-4 * np.pi, 4 * np.pi, 100)
z = np.linspace(-2, 2, 100)
r = z**2 + 1
x = r * np.sin(theta)
y = r * np.cos(theta)

# Main 3D plot
axes[0].plot(x, y, z) # The label is redundant: label='Parametric curve'
axes[0].set(
    xlabel='X', ylabel='Y', zlabel='Z',
    #title='Parametric curve'
)
axes[0].set_title(
    'Parametric curve',
    color='crimson',
    fontweight='bold',
    fontsize=14,
)
#axes1.legend(loc='lower left')
axes[0].xaxis.label.set_color('orangered')
axes[0].yaxis.label.set_color('orangered')
axes[0].zaxis.label.set_color('orangered')

# 2D projection subplot
axes[1].set_facecolor('white')
axes[1].plot(x, y, color='blue')
axes[1].set(xlabel='X', ylabel='Y') #, title='XY projection')
axes[1].set_title(
    'XY projection',
    color='blue',
    fontweight='normal',
    fontsize=10,
)
axes[1].set_aspect('equal', adjustable='box')
axes[1].grid(alpha=0.3)
axes[1].tick_params(axis='both', labelcolor='blue')
axes[1].xaxis.label.set_color('blue')
axes[1].yaxis.label.set_color('blue')


# 2D projection subplot
axes[2].set_facecolor('white')
axes[2].plot(y, z, color='orange')
axes[2].set(xlabel='Y', ylabel='Z') #, title='YZ projection')
axes[2].set_title(
    'YZ projection',
    color='orange',
    fontweight='normal',
    fontsize=10
)
axes[2].set_aspect('equal', adjustable='box')
axes[2].grid(alpha=0.3)
axes[2].tick_params(axis='both', labelcolor='orange')
axes[2].xaxis.label.set_color('orange')
axes[2].yaxis.label.set_color('orange')

# XY projection border
for spine in axes[1].spines.values():
    spine.set_edgecolor('blue')
    spine.set_linewidth(2)

# YZ projection border
for spine in axes[2].spines.values():
    spine.set_edgecolor('orange')
    spine.set_linewidth(2)

axes[1].set_facecolor('aliceblue')
axes[2].set_facecolor('cornsilk')

plt.show()

fig.savefig('./images/week_1/figure_parametric_subplots_1.png',bbox_inches='tight')
````
