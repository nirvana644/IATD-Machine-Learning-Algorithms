<h1>2. Creating figure in Matplotlib</h1>
With this library you can create figure objects and then just call methods or attributes of that object. This approach could help when you have multiple plots on it and treating them as individual object provides more flexibility.

```python
# Create Figure (empty canvas)
fig = plt.figure()

# Add set of axes to figure, practice this by changing the left, bottom, width and height to some other 
#numbers in the range of 0 and 1
axes = fig.add_axes([0, 0, 1, 1]) # left, bottom, width, height (range 0 to 1)

# Plot on that set of axes
axes.plot(x, y)

plt.show()
```

```python
import numpy as np
a = np.linspace(0,10,11)
b = a ** 2
```

```python
a
```

```python
b
```

```python
# Create Figure (empty canvas)
fig = plt.figure()

# Add set of axes to figure, practice this by changing the left, bottom, width and height to some other 
#numbers in the range of 0 and 1
axes = fig.add_axes([0, 0, 1, 1]) # left, bottom, width, height (range 0 to 1)

# Plot on that set of axes
axes.plot(a, b)

plt.show()
```

```python
# Create a blank canvas for figure
fig = plt.figure()

#now add two axes one for bigger and other for smaller plot/figure
axes1 = fig.add_axes([0, 0, 1, 1]) # Bigger figure
axes2 = fig.add_axes([0.2, 0.2, 0.5, 0.5]) # Smaller figure

# Bigger Figure Axes 1
axes1.plot(a, b)

# Use set_ to add to the axes figure
axes1.set_xlabel('X')
axes1.set_ylabel('Y')
axes1.set_title('Bigger plot')

# Plot Figure Axes 2
axes2.plot(a,b)
axes2.set_xlabel('A')
axes2.set_ylabel('B')
axes2.set_title('Smaller plot');
```

```python
# bbox_inches ='tight' automatically makes sure the bounding box is correct
fig.savefig('./images/week_1/figure.png',bbox_inches='tight')
```

<h1> Creating Sub Plots in Matplotlib </h1>

The plt.subplots() object acts as an automatic axis manager making multiple plots side by side easy.

So lets see the syntax and understand it further.

```python
# it is similar to plt.figure() except the use of tuple unpacking to grab fig and axes
fig, axes = plt.subplots()

# Again use the axes object to add stuff to plot
axes.plot(x, y, 'r')
axes.set_xlabel('x')
axes.set_ylabel('y')
axes.set_title('title'); #; hides Out[]
```

## Adding rows and columns

Then you can specify the number of rows and columns when creating the subplots() object:

```python
# Empty canvas of 1 by 2 subplots
fig, axes = plt.subplots(nrows=1, ncols=2)
```

```python
# Axes is an array of axes to plot on
axes
```

```python
axes.shape
```

```python
# Empty canvas of 2 by 2 subplots
fig, axes = plt.subplots(nrows=2, ncols=2)
```

```python
fig,axes = plt.subplots(nrows=1,ncols=2)

axes[0].plot(a,b)
axes[1].plot(x,y)
```

```python
fig,axes = plt.subplots(nrows=2,ncols=2)


axes[0][0].plot(a,b)
axes[0][1].plot(x,y)
axes[1][0].plot(b,a)
axes[1][1].plot(y,x)
```
