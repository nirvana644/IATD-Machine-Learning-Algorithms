### Matrix Multiplication: `belief @ A`

**Given:**

$$
\text{belief} = \begin{bmatrix} 0.23 & 0.77 \end{bmatrix}, \quad A = \begin{bmatrix} 0.8 & 0.2 \\ 0.3 & 0.7 \end{bmatrix}
$$

**Dimensions:**
* `belief`: $1 \times 2$ row vector
* `A`: $2 \times 2$ transition matrix
* **Result:** $1 \times 2$ row vector

---

#### Step-by-Step Calculation

1. **First entry (Row 1 \(\times\) Column 1):**
   $$(0.23 \times 0.8) + (0.77 \times 0.3) = 0.184 + 0.231 = 0.415$$

2. **Second entry (Row 1 \(\times\) Column 2):**
   $$(0.23 \times 0.2) + (0.77 \times 0.7) = 0.046 + 0.539 = 0.585$$

**Result:**
$$\text{belief} @ A = \begin{bmatrix} 0.415 & 0.585 \end{bmatrix}$$

*(Check: (0.415 + 0.585 = 1.0), confirming the probabilities sum to 100%.)*
