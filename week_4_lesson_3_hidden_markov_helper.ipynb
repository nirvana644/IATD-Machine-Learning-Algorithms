### Matrix Multiplication: `belief @ A`

**Given:**

$$
\text{belief} = \begin{bmatrix} 0.23 & 0.77 \end{bmatrix}, \quad A = \begin{bmatrix} 0.8 & 0.2 \\ 0.3 & 0.7 \end{bmatrix}
$$

**Dimensions:**
* `belief`: $1 \times 2$ row vector
* `A`: $2 \times 2$ transition matrix
* **Result:** $1 \times 2$ row vector

---

#### Step-by-Step Calculation

1. **First entry (Row 1 $\times$ Column 1):**
   $$(0.23 \times 0.8) + (0.77 \times 0.3) = 0.184 + 0.231 = 0.415$$

2. **Second entry (Row 1 $\times$ Column 2):**
   $$(0.23 \times 0.2) + (0.77 \times 0.7) = 0.046 + 0.539 = 0.585$$

**Result:**
$$\text{belief} @ A = \begin{bmatrix} 0.415 & 0.585 \end{bmatrix}$$

*(Check: (0.415 + 0.585 = 1.0), confirming the probabilities sum to 100%.)*

---

````python
p_stay = 0.95
A_exercise_four = np.array([
    [p_stay,     1 - p_stay],  # Sunny → [Sunny, Rainy]
    [1 - p_stay, p_stay]       # Rainy → [Sunny, Rainy]
])

print(A_exercise_four)
````

````python
belief = np.array([0.90000, 0.10000])

print(f"A[0, 0] = {A_exercise_four[0, 0]}")
print(f"A[1, 0] = {A_exercise_four[1, 0]}")
print(f"belief[0] = {belief[0]}")
````

````python
column = 0
first_value = belief[0] * A_exercise_four[0, column] + belief[1] * A_exercise_four[1, column]
column = 1
second_value = belief[0] * A_exercise_four[0, column] + belief[1] * A_exercise_four[1, column]

print(f"[{first_value:.3f}, {second_value:.3f}]")
print(f"Sum of values: {first_value + second_value:.3f}")
````

````python
day_5_sunny = 0.77629
day_5_rainy = 0.22371

probability_sunny_to_sunny = A[0,0]
probability_sunny_to_rainy = A[0,1]
probability_rainy_to_rainy = A[1,1]
probability_rainy_to_sunny = A[1,0]
print(f"Probability of sunny to sunny: {probability_sunny_to_sunny}")
print(f"Probability of sunny to rainy: {probability_sunny_to_rainy}")
print(f"Probability of rainy to sunny: {probability_rainy_to_sunny}")

print(f"Sum of probabilities day 5: {day_5_sunny + day_5_rainy:.2%}")

prob_day_6_sunny = (
        day_5_sunny * probability_sunny_to_sunny +
        day_5_rainy * probability_rainy_to_sunny)
prob_day_6_rainy = (
        day_5_rainy * probability_rainy_to_rainy +
        day_5_sunny * probability_sunny_to_rainy)

print(f"\nProbability day 6 is sunny: {prob_day_6_sunny:.2%}")
print(f"Probability day 6 is rainy: {prob_day_6_rainy:.2%}")
print(f"Sum of probabilities day 6: {prob_day_6_sunny + prob_day_6_rainy:.2%}")

prob_day_7_sunny = (
        prob_day_6_sunny * probability_sunny_to_sunny +
        prob_day_6_rainy * probability_rainy_to_sunny)
prob_day_7_rainy = (
        prob_day_6_rainy * probability_rainy_to_rainy +
        prob_day_6_sunny * probability_sunny_to_rainy)

print(f"\nProbability day 7 is sunny: {prob_day_7_sunny:.2%}")
print(f"Probability day 7 is rainy: {prob_day_7_rainy:.2%}")
print(f"Sum of probabilities day 7: {prob_day_7_sunny + prob_day_7_rainy:.2%}")

matrix_after_day_5 = np.array([day_5_sunny, day_5_rainy])
print(f"\nMatrix shape: {matrix_after_day_5.shape}")
prob_day_7_matrix = matrix_after_day_5 @ A @ A
print(f"Matrix after day 7: {prob_day_7_matrix}")
assert np.allclose(prob_day_7_matrix, np.array([prob_day_7_sunny, prob_day_7_rainy]))
print("Verification succeeded")

run_index = 0
sunny_index = 0
rainy_index = 1
prob_run_cond_given_sunny = B[sunny_index,run_index]
prob_run_cond_given_rainy = B[rainy_index,run_index]
prob_run_and_sunny = prob_run_cond_given_sunny * prob_day_7_sunny
prob_run_and_rainy = prob_run_cond_given_rainy * prob_day_7_rainy

print(f"\nProbability of run and sunny: {prob_run_and_sunny:.2%}")
print(f"Probability of run and rainy: {prob_run_and_rainy:.2%}")
print(f"Sum of probabilities for run: {prob_run_and_sunny + prob_run_and_rainy:.2%}")

office_index = 1
prob_office_cond_given_sunny = B[sunny_index,office_index]
prob_office_cond_given_rainy = B[rainy_index,office_index]
prob_office_and_sunny = prob_office_cond_given_sunny * prob_day_7_sunny
prob_office_and_rainy = prob_office_cond_given_rainy * prob_day_7_rainy

print(f"\nProbability of office and sunny: {prob_office_and_sunny:.2%}")
print(f"Probability of office and rainy: {prob_office_and_rainy:.2%}")
print(f"Sum of probabilities for office: {prob_office_and_sunny + prob_office_and_rainy:.2%}")

clean_index = 2
prob_clean_cond_given_sunny = B[sunny_index,clean_index]
prob_clean_cond_given_rainy = B[rainy_index,clean_index]
prob_clean_and_sunny = prob_clean_cond_given_sunny * prob_day_7_sunny
prob_clean_and_rainy = prob_clean_cond_given_rainy * prob_day_7_rainy

print(f"\nProbability of clean and sunny: {prob_clean_and_sunny:.2%}")
print(f"Probability of clean and rainy: {prob_clean_and_rainy:.2%}")
print(f"Sum of probabilities for clean: {prob_clean_and_sunny + prob_clean_and_rainy:.2%}")

print(f"\nSum of probabilities for run, office, clean: {prob_run_and_sunny + prob_run_and_rainy+ prob_office_and_sunny + prob_office_and_rainy+prob_clean_and_sunny + prob_clean_and_rainy:.2%}")


prob_activity_day_7_matrix = prob_day_7_matrix @ B
print(f"\nActivity matrix after day 7: {prob_activity_day_7_matrix}")
assert np.allclose(prob_activity_day_7_matrix, np.array([prob_run_and_sunny + prob_run_and_rainy, prob_office_and_sunny + prob_office_and_rainy, prob_clean_and_sunny + prob_clean_and_rainy]))
print("Verification succeeded")
````
